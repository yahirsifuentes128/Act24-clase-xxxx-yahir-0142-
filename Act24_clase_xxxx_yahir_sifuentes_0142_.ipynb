#### 📝 Descripción del Problema: Modelado de una Imprenta

Contexto:
Se requiere desarrollar un sistema que permita automatizar la gestión de diversos talleres de artes gráficas. Para ello, se utiliza la Programación Orientada a Objetos para crear un modelo digital de una Imprenta.

Definición de la Clase:
La clase Imprenta actúa como un molde genérico que define tanto las características físicas y operativas como los servicios que el negocio ofrece.

```python
class Imprenta:
    def __init__(self, nombre, capacidad, papel, maquinaria):
        # Atributos: Los datos que definen a tu imprenta
        self.nombre = nombre
        self.capacidadImpresion = capacidad
        self.tiposPapel = papel
        self.maquinaria = maquinaria

    # Funciones: Las acciones que puede realizar
    def disenar(self):
        print(f"\n--- 🎨 Proceso en {self.nombre} ---")
        print(f"Diseñando arte digital para imprimir en papel {self.tiposPapel}...")

    def imprimir(self):
        print(f"Ejecutando impresión de {self.capacidadImpresion} unidades usando: {self.maquinaria}.")

    def encuadernar(self):
        print(f"Finalizando: El trabajo ha sido encuadernado profesionalmente.")
        print("-" * 30)

# --- Bloque Interactivo ---
print("🏭 BIENVENIDO AL CREADOR DE IMPRENTAS 🏭\n")

# Función para registrar datos manualmente
def crear_imprenta_interactiva(numero):
    print(f"Configurando Imprenta #{numero}:")
    nom = input("¿Cuál es el nombre de la imprenta? ")
    cap = input("¿Cuál es su capacidad de impresión (ej. 5000)? ")
    pap = input("¿Qué tipo de papel usa principalmente? ")
    maq = input("¿Qué maquinaria utiliza? ")

    # Aquí se crea el objeto con los datos que tú ingresaste
    return Imprenta(nom, cap, pap, maq)

# 1. Creando la Imprenta 1 con tus datos
imprenta1 = crear_imprenta_interactiva(1)

# 2. Creando la Imprenta 2 con tus datos
print("\n")
imprenta2 = crear_imprenta_interactiva(2)

# --- Demostración de las Funciones ---
print("\n🚀 ACTIVANDO LOS OBJETOS CREADOS...")

# Ejecutamos las funciones para la Imprenta 1
imprenta1.disenar()
imprenta1.imprimir()
imprenta1.encuadernar()

# Ejecutamos las funciones para la Imprenta 2
imprenta2.disenar()
imprenta2.imprimir()
imprenta2.encuadernar()
```
